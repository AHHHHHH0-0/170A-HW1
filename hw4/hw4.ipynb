```python
import pandas as pd
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier


SEED = 67
```

```python
with open('Default.csv', 'r') as file:
    data = pd.read_csv(file)

print(f"First 5 rows of the dataset: \n{data.head()}\n")
print(f"Shape of the dataset: \n{data.shape}\n")
print(f"Null values: \n{data.isnull().sum()}\n")
```

```python
# 10-fold CV
ten_fold_cv = KFold(n_splits=10, shuffle=True, random_state=SEED)
```

```python
# Income and balance
X1 = data[['income', 'balance']]
y1 = (data['default'] == 'Yes').astype(int)

# Logistic regression model
lr_model1 = LogisticRegression(random_state=SEED)
cv_scores1 = cross_val_score(lr_model1, X1, y1, cv=ten_fold_cv, scoring='accuracy') # default 0.5 threshold 

print(f"10-fold CV test error with income and balance: \n{1 - cv_scores1.mean():.4f}")
```

```python
# Income, balance, and student
X2 = data[['income', 'balance', 'student']]
X2.loc[:, 'student'] = X2['student'].map({'Yes': 1, 'No': 0})
y2 = (data['default'] == 'Yes').astype(int)

# Logistic regression model
lr_model2 = LogisticRegression(random_state=SEED)
cv_scores2 = cross_val_score(lr_model2, X2, y2, cv=ten_fold_cv, scoring='accuracy')

print(f"10-fold CV test error with income, balance, and student: \n{1 - cv_scores2.mean():.4f}")
```

Part C:
Including a dummy variable for student status does NOT lead to a reduction of test error rate, since it increased from 2.64% (without student) to 2.66% (with student).


```python
# Convert to numeric
X3 = X2.astype(float)

# XGBoost model
xgb_model = XGBClassifier(random_state=SEED)
cv_scores_xgb = cross_val_score(xgb_model, X3, y2, cv=ten_fold_cv, scoring='accuracy')

print(f"10-fold CV test error with XGBoost: \n{1 - cv_scores_xgb.mean():.4f}")
```

Part D: It seems that Logistic Regression (0.0266) acheived a slightly lower 10-fold CV error than XGBoost (0.0313). This may be because we only have 3 features, and XGBoost thrives on more complex interactions between many features.
